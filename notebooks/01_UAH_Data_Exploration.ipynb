# UAH-DriveSet Data Exploration

## Objective

The goal of this notebook is to explore the UAH-DriveSet dataset before building any machine learning models.

This notebook focuses on:

- Understanding the folder structure
- Understanding the available sensor files
- Exploring file formats
- Identifying labels
- Understanding the relationship between different sensors

No machine learning model will be trained in this notebook.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset_path = "/content/drive/MyDrive/TOGG_Driver_Risk_Project/data/raw/UAH-DRIVESET-v1"

print(dataset_path)
```

```python
os.listdir(dataset_path)
```

```python
drivers = sorted([
    item
    for item in os.listdir(dataset_path)
    if os.path.isdir(os.path.join(dataset_path, item))
    and item.startswith("D")
])

print(f"Number of drivers: {len(drivers)}")
print(drivers)
```

```python
trip_summary = {}

for driver in drivers:

    driver_path = os.path.join(dataset_path, driver)

    trips = sorted([
        trip
        for trip in os.listdir(driver_path)
        if os.path.isdir(os.path.join(driver_path, trip))
    ])

    trip_summary[driver] = len(trips)

    print(f"{driver}: {len(trips)} trips")
```

```python
driver_path = os.path.join(dataset_path, "D1")

trips = sorted([
    trip
    for trip in os.listdir(driver_path)
    if os.path.isdir(os.path.join(driver_path, trip))
])

print("Trips for Driver D1:\n")

for trip in trips:
    print(trip)
```

```python
trip_path = os.path.join(driver_path, trips[0])

files = sorted(
    [
        file
        for file in os.listdir(trip_path)
    ]
)

print(f"Trip: {trips[0]}\n")
print(f"Number of files: {len(files)}\n")

for file in files:
    print(file)
```
