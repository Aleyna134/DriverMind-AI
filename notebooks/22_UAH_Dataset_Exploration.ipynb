```python
def inspect_file(file_name, n_lines=10):

    path = os.path.join(
        trip_path,
        file_name
    )

    print("="*70)
    print(file_name)
    print("="*70)

    print("\nFirst Lines:\n")

    with open(path, "r") as f:

        for i in range(n_lines):

            line = f.readline()

            if not line:
                break

            print(line.rstrip())

    print("\nFile Size:")

    print(
        round(os.path.getsize(path)/1024,2),
        "KB"
    )

    print("\n")
```
