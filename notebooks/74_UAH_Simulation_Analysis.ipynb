```python
from pathlib import Path
import numpy as np

PROJECT_PATH = Path("/content/drive/MyDrive/UAH_Project")

data = np.load(
    PROJECT_PATH /
    "datasets" /
    "processed" /
    "uah_dataset_window160.npz"
)

X = data["X"]

print(X.shape)
```

```python
feature_idx = {
    "acc_x":0,
    "acc_y":1,
    "acc_z":2,
    "roll":3,
    "pitch":4,
    "yaw":5,
    "speed":6,
    "heading":7,
    "lane_offset":8,
    "phi":9,
    "front_distance":10,
    "relative_speed":11,
    "vehicle_state":12
}

simulation_vectors = X[:,0,[
    feature_idx["speed"],
    feature_idx["heading"],
    feature_idx["lane_offset"],
    feature_idx["front_distance"],
    feature_idx["relative_speed"],
    feature_idx["vehicle_state"],
    feature_idx["acc_z"]
]]

simulation_vectors.shape
```

```python
np.save(
    PROJECT_PATH /
    "datasets" /
    "processed" /
    "simulation_vectors.npy",
    simulation_vectors
)

print("Saved.")
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

simulation_vectors_scaled = scaler.fit_transform(simulation_vectors)

np.save(
    PROJECT_PATH /
    "datasets" /
    "processed" /
    "simulation_vectors_scaled.npy",
    simulation_vectors_scaled,
)

print("Scaled vectors saved.")
```

```python
import joblib

joblib.dump(
    scaler,
    PROJECT_PATH /
    "datasets" /
    "processed" /
    "simulation_scaler.pkl"
)

print("Scaler saved.")
```
